## Exercice 1 : Calculer la superficie d'un quartier

### Contexte
Nous voulons calculer la superficie du quartier "West Village" en utilisant la fonction `ST_Area`.

### Données
La table `nyc_neighborhoods` contient les géométries des quartiers.

### Solution
Utilisez la fonction `ST_Area` pour calculer la superficie.

```python
# Calcul de la superficie du quartier "West Village"
query = """
SELECT ST_Area(geom)
FROM nyc_neighborhoods
WHERE name = 'West Village';
"""
df = pd.read_sql(query, engine)
df
```

## Exercice 2 : Calculer la longueur d'une rue

### Contexte
Nous voulons calculer la longueur de la rue "Pelham St" en utilisant la fonction `ST_Length`.

### Données
La table `nyc_streets` contient les géométries des rues.

### Solution
Utilisez la fonction `ST_Length` pour calculer la longueur.

```python
# Calcul de la longueur de la rue "Pelham St"
query = """
SELECT ST_Length(geom), geom
FROM public.nyc_streets
where name = 'Pelham St';
"""
df = pd.read_sql(query, engine)
df
```

## Exercice 4 : Calculer la longueur totale des rues de New York

### Contexte

### Données

Utilisez la fonction `ST_Length` pour additionner la longueur de toutes les rues.

## Exercice 5 : Trouver la station de métro la plus à l'ouest

### Contexte
Nous voulons identifier la station de métro la plus à l'ouest.

### Données
La table `nyc_subway_stations` contient les géométries des stations.

### Solution
Utilisez la fonction `ST_X` pour extraire la longitude et trouvez la station avec la valeur minimale.

```python
# Station de métro la plus à l'ouest
query = """
SELECT ST_X(geom) AS X, name
FROM public.nyc_subway_stations
order by X asc
LIMIT 1;

"""
df = pd.read_sql(query, engine)
df
```

## Exercice 6 : Trouver les rues à proximité de la station "Broad Street"

### Contexte
Nous voulons identifier les rues situées à moins de 10 mètres de la station "Broad Street".

### Données
La table `nyc_streets` contient les géométries des rues, et la table `nyc_subway_stations` contient les géométries des stations.

### Solution
Utilisez la fonction `ST_DWithin` pour sélectionner les rues proches de la station.

```python
# Rues situées à moins de 10 mètres de la station "Broad Street"
query = """
SELECT s.name
FROM public.nyc_streets s
JOIN public.nyc_subway_stations st
ON ST_DWithin(s.geom, st.geom, 10)
WHERE st.name = 'Broad St';

"""
df = pd.read_sql(query, engine)
df
```

## Exercice 7 : Localiser un point dans un quartier et un arrondissement

### Contexte
Dans quel quartier et arrondissement se trouve le point `POINT(586782 4504202)` ?

### Données
La table `nyc_neighborhoods` contient les géométries des quartiers, et la table `nyc_census_blocks` contient les géométries des arrondissements.

### Solution
Utilisez la fonction `ST_Intersects` pour déterminer dans quel quartier et arrondissement ce point se situe.

```python
# Localisation du point dans un quartier et un arrondissementp±
query = """
SELECT n.boroname AS quartier, c.boroname AS arrondissement
FROM nyc_neighborhoods n, nyc_census_blocks c
WHERE ST_Intersects(n.geom, ST_GeomFromText('POINT(586782 4504202)', 32618))
AND ST_Intersects(c.geom, ST_GeomFromText('POINT(586782 4504202)', 32618));
"""
df = pd.read_sql(query, engine)
df
```

## Exercice 8 : Calculer la distance entre deux stations

### Contexte
À quelle distance se trouvent "Cortlandt St (R,W) Manhattan" et "Baychester Ave (5) Bronx" ?

### Données
La table `nyc_subway_stations` contient les géométries des stations.

### Solution
Utilisez la fonction `ST_Distance` pour calculer la distance entre les géométries des stations.

```python
# Distance entre "Columbus Cir" et "Fulton Ave"
query = """
SELECT ST_Distance(
  (SELECT geom FROM nyc_subway_stations WHERE long_name = 'Cortlandt St (R,W) Manhattan'),
  (SELECT geom FROM nyc_subway_stations WHERE long_name = 'Baychester Ave (5) Bronx')
) AS distance;
"""
df = pd.read_sql(query, engine)
df
```

## Exercice 9 : Trouver la station de métro dans "Little Italy" et sa ligne

### Contexte
Quelle station de métro se trouve dans le quartier "Little Italy" ? Sur quelle ligne de métro se trouve-t-elle ?

### Données
La table `nyc_subway_stations` contient les géométries des stations, et la table `nyc_neighborhoods` contient les géométries des quartiers.

### Solution
Utilisez une jointure spatiale pour identifier la station et sa ligne.

```python
# Station de métro dans "Little Italy" et sa ligne
query = """
SELECT s.name AS station
FROM nyc_subway_stations s
JOIN nyc_neighborhoods n ON ST_Within(s.geom, n.geom)
WHERE n.name = 'Little Italy';
"""
df = pd.read_sql(query, engine)
df
```
