### Étape 2 : Charger les limites des communes dans GeoPandas
Utilisez GeoPandas pour lire les limites des communes depuis un GeoPackage.

#### Instructions
1. Identifiez le fichier GeoPackage contenant les limites des communes (par exemple, `swissboundaries.gpkg`).
2. Chargez les données dans un GeoDataFrame en spécifiant la couche correspondante.
3. Affichez les premières lignes pour vérifier le contenu.

````python
!pip install geopandas
import geopandas as gpd

# Chemin vers le fichier GeoPackage
geopackage_path = "swissboundaries/swissboundaries.gpkg"  # Remplacez par le chemin réel

# Charger les données depuis la couche "communes"
gdf = gpd.read_file(geopackage_path, layer="communes")

# Afficher les premières lignes
print(gdf.head())
````

### Étape 3 : Sauvegarder les données dans une base PostGIS
Utilisez GeoPandas pour exporter les données vers une base de données PostGIS.

#### Instructions
1. Configurez les paramètres de connexion à la base de données PostGIS.
2. Utilisez la méthode `to_postgis` de GeoPandas pour sauvegarder les données.
3. Vérifiez que les données ont bien été insérées dans la base.

#### Remarque
Assurez-vous que la base de données est configurée pour accepter les connexions et que l'extension PostGIS est activée.

````python
from sqlalchemy import create_engine

# Paramètres de connexion
host = "localhost"
database = "cfgeo"
user = "postgres"
password = "postgres"
table_name = "communes"

# Créer une connexion à la base de données
engine = create_engine(f"postgresql://{user}:{password}@{host}/{database}")

# Sauvegarder les données dans PostGIS
gdf.to_postgis(table_name, engine, if_exists="replace")

print(f"Données sauvegardées dans la table '{table_name}' de la base '{database}'.")
````
